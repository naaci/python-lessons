```python
from socket import *
```

## Creating A Server Socket

```python
server = socket()
server.setsockopt(SOL_SOCKET, SO_REUSEADDR, 1)
```

```python
server.bind(('localhost',80))
server.listen()
```

```python
c1, *_ = server.accept()
```

```python
print(c1.recv(4096).decode())
```

```python
c1.send("""HTTP/1.1 404 Not Found
Content-Length: 0

""".encode())
```

## Creating A Client Socket

```python
client = socket()
```

```python
client.connect(('localhost',80))
```

```python
client.send("GET / HTTP/1.1\r\n\r\n".encode())
```

```python
print(c1.recv(4096).decode())
```

```python
c1.send("""HTTP/1.1 404 Not Found
Content-Length: 0

""".encode())
```

```python
c1.close()
```

```python
server.close()
```

```python
client.close()
```
