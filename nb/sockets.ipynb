```python
from socket import *
```

```python
s = socket()
```

```python
s.bind(('localhost',8000))
s.listen()
```

```python
c1, *_ = s.accept()
```

```python
print(c1.recv(4096).decode())
```

```python
c1.send("""HTTP/1.1 404 Not Found
Content-Length: 0

""")
```

```python
c1.close()
```

```python
s.close()
```
