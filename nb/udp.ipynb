# UDP Protocol Example
In this example a simple UDP/IP server and client are implemented. 
The server listens on `localhost` and translates any string sent by the client to the uppercase letters.

```python
from socket import *
```

## Creating A UDP Server Socket

```python
server_addr = 'localhost', 12001
```

```python
server_socket = socket(AF_INET, SOCK_DGRAM)
```

Here `AF_INET` means the adress family is IP and `SOCK_DGRAM` means that the transport protocol is UDP.

```python
server_socket.bind(server_addr)
```

## Creating A Client Socket

```python
client_socket = socket(AF_INET, SOCK_DGRAM)
```

```python
message = 'hello world!'
```

```python
client_socket.sendto(encoded_message := message.encode(), server_addr)
```

```python
encoded_message
```

```python
recieved_message, client_address = server_socket.recvfrom(2048)
```

```python
recieved_message
```

```python
decoded_message = recieved_message.decode()
```

```python
server_socket.sendto(decoded_message.upper().encode(), client_address)
```

```python
recieved_modified_message, _ = client_socket.recvfrom(2048)
```

```python
print(recieved_modified_message.decode())
```

```python
client_socket.close()
server_socket.close()
```
