# Python Dictionary Objects
Dictionaries are simple mappings that associate some hashable objects to some objects.

To create a dictionary in Python, write some `key:value` pairs between braces seperating them with `,`.
Here 
- `key` must be a hashable Python object like: string, integer, floating point number, or tuple and 
- `value` is any Python object.

```python
d = {
    'name':'Ali',
    'surname':'Cin',
    'age':17,
    'address':{
     'country':'Türkiye',
     'address': 'Antalya',
     },
    'friends':['Ahmet','Mehmet'],
    17: 'ok'
    }
```

Dictionary keys must be unique and hashable.
But values can be any Python object so they can be nested.

You can create a dictionary with an iterable of `key,value` pairs using `dict()` function.

```python
dict(((3,9),(7,49),(1,1)))
```

Or you can create a dictionary with comprehension syntax.

```python
{x:x**2 for x in range(10) if x & 1}
```

In comprehension `if` part is optional.

```python
{x: x**2 for x in range(10)}
```

## Empty Dictionary
Empty dictionary can be created by `{}` or `dict()`.

## Accessing The Items of Dictionary

```python
d['address']['country']
```

```python
d[17]
```

To test if a `key` exists in a dictionary use `in` keyword.

```python
'job' in d
```

If the `key` does not exist Python will yield a `KeyError`.
To prevent this, you can use `.get(key)` method.
`.get(key)` method gives `None` if the key is not in the dictionary.

```python
d.get('job')
```

## Modifying The Dictionary

You can change the values and also add new `key:value` pairs to a dictionary.

```python
d['age'] = 18
d['gender'] = 'male'
d
```

```python
d.update({'age':19})
d
```

Use `.setdefault(key,value)` to add a `key:value` pair in to the dictionary if the `key` does not exists. If it is already in the dictionary this method gets that value.

```python
d.setdefault('friends',[])
```

To remove a `key:value` pair use `del` keyword.

```python
del d[17]
d
```

The `.clear()` method remove all items from the dictioanry.
And the `.copy()` method makes a copy of the dictionary.
