# TCP Protocol Example
In this example a simple TCP/IP server and client are implemented. 
The server listens on `localhost` and translates any string sent by the client to the uppercase letters.

```python
from socket import *
```

## Creating A TCP Server Socket

```python
server_addr = 'localhost', 12001
```

```python
server_socket = socket(AF_INET, SOCK_STREAM)
```

Here `AF_INET` means the adress family is IP and `SOCK_STREAM` means that the transport protocol is TCP.

```python
# server_socket.setsockopt(SOL_SOCKET, SO_REUSEADDR, 1)
```

```python
server_socket.bind(server_addr)
```

```python
server_socket.listen()
```

## Creating A Client Socket

```python
client_socket = socket(AF_INET, SOCK_STREAM)
```

```python
client_socket.connect(server_addr)
```

```python
client_socket.send('''GET / HTTP/1.1

'''.encode())
```

```python
connection1, client_address = server_socket.accept()
```

```python
print(connection1.recv(4096).decode())
```

```python
connection1.send('''HTTP/1.1 404 Not Found
Content-Length: 0

'''.encode())
```

```python
connection1.close()
```

```python
client_socket.close()
server_socket.close()
```
