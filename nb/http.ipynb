# HTTP Protocol

```python
from socket import *
```

```python
s = socket()
```

```python
s.connect(("www.example.com",80))
```

```python
s.send(b"""GET / HTTP/1.0
host: www.example.com

""")
```

```python
print(s.recv(4086).decode())
```

## HTTPS Protocol

```python
domain = "www.alanyauniversity.edu.tr"
```

```python
import ssl
```

```python
s = socket()
```

```python
context = ssl.create_default_context()
ss = context.wrap_socket(s, server_hostname=domain)
```

```python
ss.connect((domain,443))
```

```python
ss.send(b"""GET / HTTP/1.1
Host: %s

""" %domain.encode())
```

```python
print(ss.recv(4086).decode())
```
