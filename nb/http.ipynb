# HTTP Protocol

```python
domain = 'www.example.com'
request = f'''GET / HTTP/1.1
Host: {domain}

'''.encode()
```

```python
from socket import *
```

```python
s = socket()
```

```python
s.connect((domain,80))
```

```python
s.send(request)
```

```python
print(s.recv(4086).decode())
```

```python
s.close()
```

## HTTPS Protocol

```python
import ssl
```

```python
s = socket()
```

```python
context = ssl.create_default_context()
ss = context.wrap_socket(s, server_hostname=domain)
```

```python
ss.connect((domain,443))
```

```python
ss.send(request)
```

```python
print(ss.recv(4086).decode())
```

```python
ss.close()
```

```python
s.close()
```
