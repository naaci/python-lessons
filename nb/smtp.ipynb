# SMTP Protocol

```python
address = 'smtp.gmail.com', 587
```

```python
from socket import *
```

```python
s = socket()
```

```python
s.connect(address)
```

```python
print(s.recv(4086).decode())
```

```python
s.sendall(f"EHLO {address[0]}\n".encode())
```

```python
print(s.recv(4086).decode())
```

```python
s.sendall(b"STARTTLS\n")
```

```python
print(s.recv(4086).decode())
```

```python
import ssl
```

```python
context = ssl.create_default_context()
ss = context.wrap_socket(s, server_hostname=address[0])
```

```python
ss.sendall(f"EHLO {address[0]}\n".encode())
```

```python
print(ss.recv(4086).decode())
```

```python
ss.close()
s.close()
```
