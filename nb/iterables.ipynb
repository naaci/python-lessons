# Iterables

```python
w = 'hello', 'world', '!'
h = 'hello'
```

## Operations on Iterables

You can get the length of a tuple by the builtin `len()` function.

```python
len(w), len(h)
```

To test if an item is in the iterable use `in` keyword.

```python
'world' in w, 'hello' in w
```

To test if an item is __not__ in the iterable use `not in` keyword.

```python
'world' not in w, 'hello' not in w
```

To count the number of occurences of an item in a iterable use `.count()` method.

```python
w.count('world'), w.count('hello')
```

## Unpacking iterables
If you need you can assign different variables to each item of an iterable.
This is called unpacking.

```python
w
```

```python
first, second, third = w
```

```python
first
```

```python
second
```

```python
third
```

Also you can do this nested if the iterable is nested.

```python
(f1, f2, f3, f4, f5), second, third = w
```

```python
f1
```

### Repacking Tuple Items
You can use `*` to unpack a tuple and
use `,` to pack to a tuple.
Here `*` prevents Python from creating a nested tuple instead of a flat one.

```python
*second, None
```

Compare this with:

```python
second, None
```
