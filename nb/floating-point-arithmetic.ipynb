# Floating Point Arithmetic

A number is _floating point number_ if it contains a `.`

```python
12.3
```

```python
.1
```

You can also use _scientific notation_ when writing floating point numbers:

```python
1.2354e3
```

An integer can be written as a floating point number by adding `.` after it.

```python
12.
```

Use `+` for addition, `-` for subtraction, `*` for multiplication and `**` for exponentation like integer arithmetic.

```python
6.4*(1.25-12.)**5.
```

Also you can use `//` and `%`

```python
2.8 // 1.2
```

```python
2.8 % 1.2
```

For (true) division use: `/`

```python
15.3 / 4.2
```

The (true) division is always done using floating point arithmetic.
Even if the result is integer.

```python
12 / 3
```

When doing arithmetic
if any of the operands is a floating point number then _floating point arithmetic_ is performed.

```python
12 + 13.
```

Floating point arithmetic is very similar to integer arithmetic
but uses different algorithms for computation.

**Warning:** Floating point arithmetic sometimes gives surprising results.
For example:

```python
0.1 * 3
```

Therefore when working on floating point numbers you should know that the result may not be exact.

Unlike integers floating point numbers can **not** be arbitrary big.

```python
2.0**1023
```

To get some floating point number `number` rounded to `n` digits use `round(number,ndigits=n)`

```python
round(12.34567, ndigits=3)
```

## `math` Module
Modules are containers for predefined functions other then the builtin ones.
To use more mathematical functions use `math` module.
One way of using `math` module is:

```python
from math import *
```

Then you can use many mathematical functions defined there.

```python
sin(2.3)
```

```python
log10(123)
```

## Complex Number Arithmetic
Python also supports complex number arithmetic.
Any complex number of the form $a+bi\in\mathbb C$ can be represented by `a+bj` in Python.

```python
((3 + 1j) * 2) ** 2
```

To get the real and imaginary part of a number use `.real` and `.imag` properties.

```python
(2 + 3j).real
```

```python
(2 + 3j).imag
```

To get the absolute value of a complex number use `abs()` function.

```python
abs(3+4j)
```

## `cmath` Module
To use the mathematical functions defined on complex numbers use `cmath` istead of `math` module.
One way of using `cmath` module is:

```python
from cmath import *
```

Then you can use mathematical functions similar to `math` module.

```python
sin(2.3-1j)
```

```python
log10(123-3j)
```
