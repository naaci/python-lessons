## Square Matrices

```python
c = np.matrix([
    [1, 2, 4],
    [2, 1, 0],
    [1, 1, 0],
])
c
```

```python
c.shape
```

```python
c ** 3
```

```python
c ** 0
```

```python
np.identity(3)
```

```python
np.zeros((2,3))
```

```python
c**-1
```

```python
c @ c ** -1
```

### Determinant

```python
np.linalg.det(c)
```

```python
np.linalg.solve(c,c[:,0])
```

```python
c @ np.matrix([[1.],
               [0.],
               [-0.]]) == c[:, 0]
```

```python
all(np.isclose(3 * np.matrix([
    [0.1],
    [0.2],
    [0.3],
]), np.matrix([[0.3],
               [0.6],
               [0.9]])))
```
