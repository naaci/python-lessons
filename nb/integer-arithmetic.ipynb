# Integer Arithmetic

For addition and subtraction use usual `+` and `-` operators.

```python
12 + 3
```

```python
15 - 12
```

For multiplication use: `*`

e.g. $12535514 \times 923136652544$

```python
12535514 * 923136652544
```

For exponentation use: `**`

e.g. $2^{1024}$

```python
2**1024
```

You can also use `pow()` function  for exponentation.

```python
pow(2, 1024)
```

**Info:** Python can hold arbitrary big integers as you can see in the previous result.

To get the absolute value of a number use `abs()` function.

```python
abs(-42)
```

To get the maximum of some numbers use `max()` function.

```python
max(1, 5, 3, 9, 2)
```

To get the minimum of some numbers use `min()` function.

```python
min(1, 5, 3, 9, 2)
```

## Modular Arithmetic
Modular arithmetic is done in Python with builtin `divmod()` function.

```python
# Get the division and modulus of 17 divided by 5
divmod(17, 5)
```

If you only need the floor division then:

```python
# the division of 17 divided by 5
17 // 5
```

If you only need the remainder then:

```python
# the remainder of 17 divided by 5
17 % 5
```

Since Python can hold really big numbers, when writing integers, you can use `_` inside integer to put a space for better readablity.
Python will ignore that `_` characters. 

```python
923_136_652_544_123_9774
```

## Operator Precedence
Operator preceedence in Python is like in mathematics.

| Operator         | Description                                           |
| --------         | -----------                                           |
| `**`             | Exponentiation                                        |
| `+`  `-`         | Unary plus, unary minus                               |
| `*` `//` `%` `/` | Multiplication, floor division, and modulus, division |
| `+` `-`          | Addition and subtraction                              |


For example:

```python
-12**3+5/7
```

is equivalent to:

```python
(-(12**3))+(5/7)
```

## Representing Characters With Integers
Any character in any alphabet can be represented by an integer.
For this we need a one-to-one mapping from the alphabet to the integers called _character set_.

- The simplest character set is ASCII which maps the English alphabet to integers between 0 and 127.
- Unicode is the most general character set in the world.
It maps all human alphabets including ancient ones like hieroglyphics and cuneiform to numbers between 0 and 1112064.
- For backward compatibility integer representatin of ASCII characters are same in unicode too.

Characters are represented by writing them between `'` or `"` in Python.

```python
# To get the integer representation of character 'A'
ord("A")
```

```python
# To get the character representation of integer 65
chr(65)
```

You can do integer arithmetic on characters.
For example:

```python
# To convert uppercase 'A' to lowercase 'a'
chr(ord("A") + 32)
```

**Warning:** This may not work for characters other than ASCII characters.

```python
chr(ord("İ") + 32)
```

## Comparison Operators
To compare integers:
- for $<$ and $>$ use:
`<`, `>`
- for $\le$ and $\ge$ use:
`<=`, `>=`
- for $=$ use:
`==`
- for $\ne$ use:
`!=`

**Warning:** `=` is used for assignment in Python. For equality check use: `==`

The result of a comparision is always one of the two special objects: `True` or `False`

```python
12 > 1, 12 >= 1, 12 <= 1, 12 == 1, 12 != 1
```

`not` makes `True` $→$ `False` and vice versa.

```python
not 12 > 1
```

Since `True` and `False` are also some Python objects they can be compared as well.

```python
(12 != 1) == (12 == 12)
```

In Python you can combine multiple comparisions in a single expression:

```python
1 < 12 < 30 > 1 <= 100
```

**Warning:** The previous example is different than the following example. 

```python
1 < 12 < (30 > 1) <= 100
```

Because when you use paranthesis, Python will evaluate the expression between paranthesis first.
So `(30 > 1)` will become `True`. Then the comparision will be done like `1 < 12 < True <= 100`.
Similarly the following expression is also different.

```python
12 != 1 == 12 == 12
```

Alternatively, you can use:
- `and` operator for $∧$ in mathematics
- `or` operator for $∨$ in mathematics

| `a`     | `b`    | `a and b` | `a or b` |
| :-:     |    :-: | :---: | :---: |
| `False` | `False`|`False`|`False`|
| `False` | `True` |`False`|`True` |
| `True`  | `False`|`False`|`True` |
| `True`  | `True` |`True` |`True` |

to combine multiple comparisions.
The precendence of all arithmetic operators and bitwise operators are higher than `not` operator then `and` operator
then `or` operator. 

```python
1 < 12 and 12 < 30 and 30 > 1 and 1 <= 100 or not True
```

is equivcalent to:

```python
((1 < 12) and (12 < 30) and (30 > 1) and (1 <= 100)) or (not True)
```
