# UPNP

## UPNP `M_SEARCH`

```python
from socket import *
```

```python
MULTICAST_ADDRESS = '239.255.255.250', 1900
```

```python
multicast_socket = socket(AF_INET, SOCK_DGRAM)
multicast_socket.settimeout(4)
multicast_socket.sendto(('''M-SEARCH * HTTP/1.1
HOST: %s:%d
MAN: "ssdp:discover"
MX: 4
ST: ssdp:all
''' % MULTICAST_ADDRESS).encode(), MULTICAST_ADDRESS)
```

```python
try:
    while True:
        print(multicast_socket.recv(512).decode())
except timeout:
    pass
```

## UPNP `B_SEARCH`

```python
from socket import *
```

```python
BROADCAST_ADDRESS = '255.255.255.255', 1990
```

```python
broadcast_socket = socket(AF_INET, SOCK_DGRAM)
broadcast_socket.settimeout(4)

broadcast_socket.setsockopt(SOL_SOCKET, SO_BROADCAST, 1)
broadcast_socket.sendto(b'''B-SEARCH * HTTP/1.1
HOST: 255.255.255.255:1990
MAN: "ssdp:discover"
MX: 4
ST: ssdp:all
''', BROADCAST_ADDRESS)
```

```python
try:
    while True:
        print(broadcast_socket.recv(512).decode())
except timeout:
    pass
```
